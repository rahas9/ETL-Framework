```python
import json
postgres_host = "bore.pub"
postgres_port = 52764
postgres_database = "sales_db"
postgres_user = "dbx_reader"
postgres_password = "shaik@123"

connection_config = {
    "host": postgres_host,
    "port": postgres_port,
    "database": postgres_database,
    "user": postgres_user,
    "password": postgres_password,
    "driver": "org.postgresql.Driver"
}

connection_json = json.dumps(connection_config)
config = json.loads(connection_json)

print("PostgreSQL configuration created successfully",type(config))

```

```python
host = config['host']
port = config["port"]
database = config["database"]
username = config["user"]
password = config["password"]
driver = config["driver"]

jdbc_url = (
    f"jdbc:postgresql://{host}:{port}/{database}?sslmode=require"
)

connection_properties = {
    "user": username,
    "password": password,
    "driver": driver
}

```

```python
def read_postgres_sql(sql_query):

    query = f"({sql_query}) AS postgres_source"

    df = spark.read.jdbc(
        url=jdbc_url,
        table=query,
        properties=connection_properties
    )

    return df



df = read_postgres_sql("""
    SELECT *
    FROM banking.customers
    LIMIT 10
""")

display(df)
```
